```python
from astropy.coordinates import SkyCoord
import astropy.units as u
import pandas as pd


def find_candidate_counterparts(optical_df, xray_df, k=1.0):
    """
    Two-stage crossmatch:
    1) search within r95
    2) if empty AND r95 < 2 arcsec → search within 2 arcsec

    Adds a column 'search_radius_flag' = 'r95' or '2'
    """

    optical_coords = SkyCoord(
        ra=optical_df["RA"].values * u.deg,
        dec=optical_df["Decl"].values * u.deg
    )

    xray_coords = SkyCoord(
        ra=xray_df["RA"].values * u.deg,
        dec=xray_df["Decl"].values * u.deg
    )

    results = []

    # 2 arcsec in degrees
    arcsec_2_deg = 2 / 3600

    for i, x_id in enumerate(xray_df.index):

        r95_deg = xray_df.loc[x_id, "r95"] * k

        # --- FIRST SEARCH: r95 ---
        radius = r95_deg * u.deg

        idx_opt, _, d2d, _ = optical_coords.search_around_sky(
            xray_coords[i], radius
        )

        if len(idx_opt) > 0:
            # Found matches in r95
            flag = "r95"

        else:
            # No matches → check if we can expand
            if r95_deg < arcsec_2_deg:
                radius = arcsec_2_deg * u.deg

                idx_opt, _, d2d, _ = optical_coords.search_around_sky(
                    xray_coords[i], radius
                )

                flag = "2"
            else:
                # Do not expand
                flag = "r95"

        # --- SAVE RESULTS ---
        if len(idx_opt) == 0:
            # No candidates at all → still record it
            results.append({
                "xray_id": x_id,
                "opt_id": None,
                "separation_arcsec": None,
                "search_radius_flag": flag,
                "n_candidates": 0
            })
        else:
            for j, sep in zip(idx_opt, d2d):
                results.append({
                    "xray_id": x_id,
                    "opt_id": optical_df.iloc[j]["Id"],
                    "separation_arcsec": sep.arcsec,
                    "search_radius_flag": flag,
                    "n_candidates": len(idx_opt)
                })

    return pd.DataFrame(results)
```
