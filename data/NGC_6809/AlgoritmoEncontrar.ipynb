```python
from matplotlib import colors
import numpy as np
import pandas as pd
import json
import math

path_xray = "C:/Users/HP/Desktop/Curso de Python/Workspace/miweb/Astronomía/data_obtained.txt"
with open(path_xray, "r") as f:
    NGC6809_sources = json.load(f)
HC_score = 0.25
MV_XR_score = 0.25
optical_score = 0.5
df = pd.DataFrame(NGC6809_sources)
df["AB"] = float(0)
df["CV"] = float(0)
df["LMXRB"] = float(0)
df
```

```python
for i in range(len(df)):
    
    HC = df.iloc[i]["Hardness_classification"]
    MV_XRAY = df.iloc[i]["Mv vs xray"]
    
    if HC == "LMXRB":
        df.loc[i, "LMXRB"] += HC_score
    elif HC == "CV":
        df.loc[i, "CV"] += HC_score
    elif HC == "CV & AB":
        df.loc[i, "CV"] += HC_score/2
        df.loc[i, "AB"] += HC_score/2
    elif HC == "AB":
        df.loc[i, "AB"] += HC_score

df
```

```python
for i in range(len(df)):
    HC = df.iloc[i]["Hardness_classification"]
    MV_XRAY = df.iloc[i]["Mv vs xray"]
    
    if MV_XRAY == "LMXRB":
        df.loc[i, "LMXRB"] += MV_XR_score
    elif MV_XRAY == "CV":
        df.loc[i, "CV"] += MV_XR_score
    elif MV_XRAY == "AB":
        df.loc[i, "AB"] += MV_XR_score

df
```

```python
for i in range(len(df)):
    HC = df.iloc[i]["Hardness_classification"]
    MV_XRAY = df.iloc[i]["Mv vs xray"]
    for j in range(3):
        pos = df.iloc[i][f"pos_{j}"]
        if pos == "bluer than MS L1":
            if df.iloc[i]["LMXRB"] > df.iloc[i]["CV"]:
                if df.iloc[i]["LMXRB"] > df.iloc[i]["AB"]:
                    df.loc[i, "LMXRB"] += (optical_score-(optical_score/6)*(j-1))/3
                else:
                    df.loc[i, "LMXRB"] += (optical_score-(optical_score/6)*(j-1))/5
            elif df.iloc[i]["LMXRB"] < df.iloc[i]["CV"]:
                if df.iloc[i]["CV"] > df.iloc[i]["AB"]:
                    df.loc[i, "CV"] += (optical_score-(optical_score/6)*(j-1))/3
                else:
                    df.loc[i, "CV"] += (optical_score-(optical_score/6)*(j-1))/5
            else:
                df.loc[i, "LMXRB"] += optical_score/6
                df.loc[i, "CV"] += optical_score/6
        elif pos == "redder than MS L1" or pos == "sub-sub-giant branch" or pos == "Red giant branch":
            if df.iloc[i]["AB"] > df.iloc[i]["CV"] and df.iloc[i]["AB"] > df.iloc[i]["LMXRB"]:
                df.loc[i, "AB"] += (optical_score+(optical_score/6)*(j-1))/3
            else:
                df.loc[i, "AB"] += (optical_score+(optical_score/6)*(j-1))/5
        elif pos == "MS" or pos == "MSTO":
            if df.iloc[i]["AB"] > df.iloc[i]["CV"] and df.iloc[i]["AB"] > df.iloc[i]["LMXRB"]:
                df.loc[i, "AB"] += optical_score/6

df 
```

```python
df["posibility_counterpart"] = df["AB"] + df["CV"] + df["LMXRB"]
df
```

```python
df["posibility_not_counterpart"] = 1 - df["posibility_counterpart"]
total_posibility = 1 - df.groupby("Xray source")["posibility_not_counterpart"].prod()

for i in range(len(df)):
    array = np.array([df.iloc[i]["AB"], df.iloc[i]["CV"], df.iloc[i]["LMXRB"]])
    sorted_array = sorted(array, reverse=True)
    if sorted_array[1] != 0:
        min_rate = sorted_array[0]/sorted_array[1]
    else:
        min_rate = 1000000000
    df.loc[i, "posibility_counterpart_weighted"] = df.iloc[i]["posibility_counterpart"]*min_rate
sum_after = df.groupby("Xray source")["posibility_counterpart_weighted"].sum()
for i in range(len(df)):
    df.loc[i, "posibility_counterpart_new"] = df.loc[i, "posibility_counterpart_weighted"] * total_posibility.loc[df.iloc[i]["Xray source"]] / sum_after.loc[df.iloc[i]["Xray source"]]
df
```

```python
for i in range(len(df)):
    if df.iloc[i]["radius"] == '2':
        df.loc[i, "posibility_counterpart_new"] *= 0.8
df
```

```python
df["AB"] *= df["posibility_counterpart_new"]/df["posibility_counterpart"]
df["CV"] *= df["posibility_counterpart_new"]/df["posibility_counterpart"]
df["LMXRB"] *= df["posibility_counterpart_new"]/df["posibility_counterpart"]
df.drop(columns = ["posibility_counterpart", "posibility_not_counterpart", "posibility_counterpart_weighted"],inplace = True)

prob_sum = df.groupby("Xray source")["posibility_counterpart_new"].sum()
for i in range(len(df)):
    df.loc[i,"total_prob"] = prob_sum.loc[df.loc[i, "Xray source"]]
df
```

```python
df["AB(%)"] = round(df["AB"] * 100, 1)
df["CV(%)"] = round(df["CV"] * 100, 1)
df["LMXRB(%)"] = round(df["LMXRB"] * 100, 1)
df["posibility_counterpart_new(%)"] = round(df["posibility_counterpart_new"] * 100, 1)
df["total_prob(%)"] = round(df["total_prob"] * 100, 1)
df_final = df[["id", "AB(%)", "CV(%)", "LMXRB(%)", "posibility_counterpart_new(%)", "total_prob(%)"]]
df_final
```
